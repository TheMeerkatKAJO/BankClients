```python

# БИБОИОТЕКИ
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.metrics import mean_squared_error
from google.colab import files
import os

# ИЗОБРАДЕНИЕ
print("Выберите файл изображения для загрузки:")
uploaded = files.upload()

filename = list(uploaded.keys())[0]
img_pil = np.array(Image.open(filename).convert("RGB"))

print(f"Загружено изображение: {filename}")
print(f"Размер: {img_pil.shape}, тип: {img_pil.dtype}")

if img_pil.shape[0] < 300 or img_pil.shape[1] < 300:
    img_pil = np.array(Image.fromarray(img_pil).resize((512, 512)))
    print(f"Изображение увеличено до: {img_pil.shape}")

# Часть А
original_shape = img_pil.shape
height, width, channels = original_shape
original_size_bytes = 3 * height * width * 1

print(f"\nHeight: {height}, Width: {width}, Channels: {channels}")
print(f"Исходный размер: {original_size_bytes} байт ({original_size_bytes/1024:.1f} КБ)")

# Часть Б
list_of_ranks = [1, 5, 10, 30, 100, min(height, width)]
print(f"\nСписок рангов: {list_of_ranks}")

# Применяем SVD к каждому из 3 каналов (R, G, B) отдельно
svd_results = []
for c in range(channels):
    channel = img_pil[:, :, c].astype(np.float64)
    U, S, Vt = np.linalg.svd(channel, full_matrices=False)
    svd_results.append((U, S, Vt))
    print(f"Канал {c}: U shape={U.shape}, S shape={S.shape}, Vt shape={Vt.shape}")

# Функция восстановления изображения для заданного ранга r
def reconstruct_image(svd_results, r, channels=3):
    """
    Восстанавливает изображение, используя первые r сингулярных чисел.
    Оптимизация: (U[:,:r] * S[:r]) @ Vt[:r,:] вместо U @ diag(S) @ Vt
    """
    reconstructed_channels = []
    for c in range(channels):
        U, S, Vt = svd_results[c]
        U_r = U[:, :r]
        S_r = S[:r]
        Vt_r = Vt[:r, :]
        channel_approx = (U_r * S_r) @ Vt_r
        reconstructed_channels.append(channel_approx)

    reconstructed = np.stack(reconstructed_channels, axis=2)
    reconstructed = np.clip(reconstructed, 0, 255).astype(np.uint8)
    return reconstructed

# Создаём список восстановленных изображений для каждого ранга
approximations = []
for r in list_of_ranks:
    img_approx = reconstruct_image(svd_results, r, channels)
    approximations.append(img_approx)
    print(f"Ранг r={r}: восстановленное изображение shape={img_approx.shape}")

# Часть В
# Оригинальное изображение
fig, ax = plt.subplots(1, 1, figsize=(6, 6))
ax.imshow(img_pil)
ax.set_title("Оригинальное изображение", fontsize=14)
ax.axis('off')
plt.tight_layout()
plt.show()

# Сжатые версии (сетка 2x3)
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for i, r in enumerate(list_of_ranks):
    compressed_size_bytes = r * (height + width + 1) * channels * 4
    compression_ratio = original_size_bytes / compressed_size_bytes

    axes[i].imshow(approximations[i])
    axes[i].set_title(f"Ранг = {r}\nСжатие в {compression_ratio:.1f} раз", fontsize=11)
    axes[i].axis('off')

plt.suptitle("Восстановление изображения методом SVD (разные ранги)", fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# Часть Г
print("\n" + "=" * 70)
print("АНАЛИЗ ЭФФЕКТИВНОСТИ СЖАТИЯ")
print("=" * 70)
print(f"Исходный размер (RGB, uint8): {original_size_bytes} байт = {original_size_bytes/1024:.1f} КБ\n")
print(f"{'Ранг r':<10} {'Сжатый размер (float32)':<25} {'Коэфф. сжатия':<15} {'MSE':<10}")
print("-" * 70)

mse_list = []
for i, r in enumerate(list_of_ranks):
    num_values_per_channel = r * (height + width + 1)
    total_values = channels * num_values_per_channel
    compressed_size_bytes = total_values * 4

    compression_ratio = original_size_bytes / compressed_size_bytes

    # Расчёт MSE между оригиналом и аппроксимацией
    mse = mean_squared_error(img_pil.flatten(), approximations[i].flatten())
    mse_list.append(mse)

    print(f"{r:<10} {compressed_size_bytes:<25} {compression_ratio:<15.2f} {mse:<10.2f}")

print(f"\nФормула сжатого размера (float32, RGB):")
print(f"  compressed_size = 3 × r × (height + width + 1) × 4 байт")
print(f"  Для r={list_of_ranks[-1]} (полное восстановление): "
      f"{3 * list_of_ranks[-1] * (height + width + 1) * 4} байт")
print(f"  Это больше исходного ({original_size_bytes} байт), т.к. float32 (4 байта) > uint8 (1 байт)")

# Проверка
print("\n" + "=" * 70)
print("АВТОПРОВЕРКА")
print("=" * 70)

# 1. Проверка размерностей
for i, r in enumerate(list_of_ranks):
    assert approximations[i].shape == original_shape, \
        f"Ошибка: размерность для ранга {r} не совпадает с исходной"

# 2. Проверка, что значения не выходят за пределы 0-255
for i, r in enumerate(list_of_ranks):
    arr = approximations[i]
    if arr.dtype != np.uint8:
        arr = np.clip(arr, 0, 255)
    assert arr.max() <= 255.1, f"Значения превышают 255 для ранга {r}"
    assert arr.min() >= -0.1, f"Значения меньше 0 для ранга {r}"

# 3. Проверка монотонности MSE
mse_list_check = []
for i in range(len(approximations)):
    mse = mean_squared_error(img_pil.flatten(), approximations[i].flatten())
    mse_list_check.append(mse)

for i in range(1, len(mse_list_check)):
    assert mse_list_check[i] <= mse_list_check[i-1] + 1e-6, \
        f"MSE не уменьшается между рангами {list_of_ranks[i-1]} и {list_of_ranks[i]}"

print("Все проверки пройдены.")
```

## Выводы

**1. При каком ранге визуальное качество перестаёт отличаться от оригинала?**
При ранге r = 100 изображение визуально почти не отличается от оригинала. Все основные детали, цвета и границы восстановлены, мелкие артефакты сжатия незаметны невооружённым глазом.

**2. Для какого ранга достигается сжатие в 10 раз?**
Для изображения 300×300 сжатие в 10 раз достигается уже при ранге r = 3 (коэффициент ~12.5) или r = 4 (коэффициент ~9.4). При таких малых рангах изображение остаётся узнаваемым по общим контурам и цветам, но теряются почти все мелкие детали, появляется сильное размытие и блочность.

**3. Почему для тёмных/светлых участков ошибка разная?**
Метод SVD лучше всего аппроксимирует плавные, однородные области . Резкие переходы, контрастные границы и мелкие детали требуют гораздо большего числа сингулярных чисел для точного описания. Поэтому при малых рангах края объектов размываются в первую очередь, а гладкие тёмные или светлые участки восстанавливаются хорошо.
